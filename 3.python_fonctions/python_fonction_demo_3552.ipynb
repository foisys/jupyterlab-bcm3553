# BCM3552 - Mise en application des fonctions tel que montré dans le cours 2 #

On peut appliquer les fonctions de deux façon différentes.

## À l'intérieur de votre code

```python
def classify_aa(aa):
    if aa in "DEKRH":
        return "chargé"
    elif aa in "STNQYC":
        return "polaire"
    elif aa in "AVLIMFWPG":
        return "hydrophobe"

def count_categories(seq):
    charged, polar, hydrophobic = 0, 0, 0
    for aa in seq:
        cat = classify_aa(aa)
        if cat == "chargé":
            charged += 1
        elif cat == "polaire":
            polar += 1
        elif cat == "hydrophobe":
            hydrophobic += 1
    return charged, polar, hydrophobic

lesLignes = []

with open("../z.misc_files/data_seq/h.sapiens.cttn.iso.a.aa.fa", "r") as aFile:
  lesLignes = aFile.readlines()

head = lesLignes[0]

lesLignes.pop(0)

seq = ""

for uneLigne in lesLignes:  
    seq = uneLigne.strip()+seq

data = count_categories(seq)

print(f"{head} : {data}")
```
