## Les décisions ne sont que des booléens

````python
aa = "K"
charged = "DEKRH"
is_charged = aa in charged
print(is_charged)
print(type(is_charged))
````

## `for` ou `while` — choisir le bon outil

Utilisez `for` lorsque vous savez combien de fois répéter (ou que vous parcourez un objet). Utilisez `while` lorsque vous répétez jusqu’à ce qu’une *condition* devienne fausse et que vous ne savez pas à l’avance combien d’itérations seront nécessaires.


## Le résultat : compter sans `.count()`

La dernière heure, `cttn_seq.count("G")` faisait cela en un appel. Aujourd’hui, construisez la boucle qui se cache derrière — car cette même boucle, avec une condition différente, peut classer chaque acide aminé selon sa chimie, plutôt que de compter une seule lettre.


````python
seq = "MWKASAGHAV"
charged_set = "DEKRH"
polar_set = "STNQYC"
hydrophobic_set = "AVLIMFWPG"

charged, polar, hydrophobic = 0, 0, 0

for aa in seq:
    if aa in charged_set:
        charged += 1
    elif aa in polar_set:
        polar += 1
    elif aa in hydrophobic_set:
        hydrophobic += 1

print(charged, polar, hydrophobic)
````

**À essayer :** classez chaque résidu à la main — suivez `seq` lettre par lettre et vérifiez que les décomptes ci-dessus correspondent à ce que vous obtiendriez manuellement.


````python
seq = "MWKASAGHAV"
for aa in seq:
    if aa in "DEKRH":
        print(aa, "chargé")
    elif aa in "STNQYC":
        print(aa, "polaire")
    elif aa in "AVLIMFWPG":
        print(aa, "hydrophobe")
````

## Le résultat


````python
print(charged, polar, hydrophobic)   # (2, 1, 7)
````

## 📋 Aide-mémoire — conditions

| Syntaxe | Ce qu’elle fait | Utilisée aujourd’hui comme |
|---|---|---|
| `if x:` | Exécute un bloc seulement si `x` est `True` | `if gc_percent > 50:` |
| `if x: ... else:` | Exécute une branche ou l’autre | `if ... else: ...` |
| `if / elif / else` | Vérifie plusieurs conditions dans l’ordre | `if / elif / else` sur `length` |
| `==, !=, <, >, <=, >=` | Opérateurs de comparaison | `aa == "D"` |
| `and / or / not` | Combine ou inverse des conditions | `not is_dna` |
| `"y" in x` | Test d’appartenance | `"D" in charged` |

## 📋 Aide-mémoire — boucles

| Syntaxe | Ce qu’elle fait | Utilisée aujourd’hui comme |
|---|---|---|
| `for i in range(n):` | Répète n fois, i = 0 .. n-1 | `for i in range(5):` |
| `for x in "abc":` | Parcourt directement les caractères d’une chaîne | `for aa in "MWKA...":` |
| `while x:` | Répète tant que `x` est `True` | `while count > 0:` |
| `break` | Quitte immédiatement la boucle | `if aa == "K": break` |
| `continue` | Passe à l’itération suivante | `if aa == "A": continue` |
