# BCM3553 - Exemple de code intégrant Bio.Entrez et Bio.Restriction #

Voici un exemple de code intégrant les concepts de recherche à distance et d'analyse de BioPython

```python
from Bio import Entrez
from Bio import SeqIO
from Bio.Restriction import *

# Identification obligatoire :-)
# Retirez le dièse pour faire fonctionner le script
#Entrez.email = "votre.nom@umontreal.ca"

handle = Entrez.esearch(db="nucleotide", term="NM_005231")
record = Entrez.read(handle)

handle = Entrez.efetch(db="nucleotide", id=record["IdList"],rettype="gb", retmode="text")

seq_record = SeqIO.read(handle, "genbank")

rb = RestrictionBatch([EcoRI, BamHI, HindIII, SalI])
result = Analysis(rb, seq_record.seq).full()
# On devrait avoir 5 sites au total, 2 pour EcoRI...
print(result)
```

```python
from Bio import Entrez
from Bio import SeqIO
from Bio.Restriction import *

# Identification obligatoire :-)
# Retirez le dièse pour faire fonctionner le script 
#Entrez.email = "votre.nom@umontreal.ca"

handle = Entrez.esearch(db="nucleotide", term="NG_029881.1")
record = Entrez.read(handle)
# On accède directement à l'identificateur avec record["IdList"] 
handle = Entrez.efetch(db="nucleotide", id=record["IdList"],rettype="gb", retmode="text")

seq_record = SeqIO.read(handle, "genbank")

rb = RestrictionBatch([EcoRI, BamHI, HindIII, SalI])
result = Analysis(rb, seq_record.seq).full()
# On devrait avoir 27 sites dont 9 sites EcoRI...
print(result)
```
