# Jupyter: comment ça fonctionne!

## Introduction

Jupyter est un environnement de développement créé spécifiquemment pour l'analyse de données en tout genre grâce à sa capacité d'interagir avec un ensemble de languages, particulièrement avec Python et R.

## Comment créer du contenu dans Jupyter?

C'est pas très complexe: on crée dans une page (un *notebook* dans le monde Jupyter) qui contiendra des *cellules* contenant soit du texte formatté (cellules de type *Markdown*) ou du code prêt à exécuter (cellules de type *Code*).

Le concept du format Markdown est très semblable à celui du format HTML (en fait le M de HTML veut dire Markdown): on écrit du texte contenant des caractères servant de balises. Ces balises sont alors lues et interprétées comme des instructions de formattage. Le format Markdown utilisé par Jupyter est capable de générer des pages de texte faciles à lire et pas trop complexe à éditer. Une feuille d'aide sur le format Markdown se trouve sur [cette page](https://www.ibm.com/docs/en/watson-studio-local/1.2.3?topic=notebooks-markdown-jupyter-cheatsheet). Une fois votre texte écrit dans la cellule, appuyez sur *Shift-Enter* pour voir le résultat :-) Une foule de modules existent pour faciliter la création de divers formes de contenu mais on va garder ça simple.

Créer du contenu de type Code n'est pas vraiment plus complexe. À la base, Jupyter comprend Python et les commandes de la console. Pour cet exemple, nous allons reprendre les exemples du cours 1 sur les commandes Unix.



## Comment interagir avec votre environnement comme si Jupyter était votre terminal

Les fichiers utilisés pour les démos se trouvent tous dans le répertoire appelé z.misc_files qui se trouve à la racine du cartable, donc un niveau plus haut que cette page. Tenez en compte lorsque vous testerez ces exemples!

Pour accéder à votre environnement via une cellule Code, il faut utiliser le caractère spécial "!" (point d'exclamation):

```python
# Vous pouvez aussi mettre des commentaires après le caractère # (dièse)
!pwd
```

```python
# Affichage des permissions
# Comment interpréter le bloc à gauche?
!ls -l
```

```python
# Usage des commandes head et tail
! head -n 2 ../z.misc_files/data_seq/cttn_teaching_multiset.fa
```

```python
# Usage des commandes head et tail
! tail -n 2 ../z.misc_files/data_seq/cttn_teaching_multiset.fa
```

```python
# Les usages de grep
# Pourquoi cette commande pourrait être importante?
! grep ">" ../z.misc_files/data_seq/cttn_teaching_multiset.fa
```

```python
# Les usages de grep
# Avec les concepts vus en classe, pourriez-vous imaginer une autre
# solution donnant le même résultat?
! grep -c ">" ../z.misc_files/data_seq/cttn_teaching_multiset.fa
```
