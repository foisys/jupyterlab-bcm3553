## Variables — Affectation

`nom = valeur` — l'affectation n'est **pas** une égalité. Règles de
nommage : lettres, chiffres, tirets bas ; ne peut pas commencer par un
chiffre ; sensible à la casse.

```python
cttn_seq = "GGAACCGGAAGTAGAGCCTGGTGCCTGGGAGCGGCTGGCGCGGCGGAATCCAGGGCCGACCCGGGCCGGACCGACCCCAGGCGGCGACGGAATCAGTCCCCAATGCCTGGAAATTCCTCATTGGATTACTGTGTTTTAAACAGAATTTCG"
print(cttn_seq)
```

**Une erreur volontaire** — les noms de variables ne peuvent pas commencer
par un chiffre :

```python
# Essayer d'exécuter cette commmande...
1cttn_seq = "GGAACC..."

# Python lit `1cttn_seq` comme une tentative de commencer un nombre (`1`)
# puis se retrouve dérouté par les lettres qui suivent immédiatement.
# Lisez l'erreur calmement — elle vous dit exactement ce qui ne va pas.
```

## Les bases des chaînes de caractères sur des données de séquence

`len()`, l'indexation `seq[0]`, le découpage `seq[0:10]`,
`.upper()` / `.lower()`, `.count()`, `+` pour la concaténation.

```python
len(cttn_seq)   # quelle est la longueur de cette séquence ?
```

```python
cttn_seq[0:10]   # les 10 premières bases
```

```python
cttn_seq.count("G")
```

```python
cttn_seq.count("C")
```

**Le résultat :** calculer le contenu GC à la main — pas de boucle, juste
des méthodes de chaînes.

```python
gc_percent = (cttn_seq.count("G") + cttn_seq.count("C")) / len(cttn_seq) * 100
print(gc_percent)
```

Vous venez de calculer un vrai pourcentage GC correct (**61,3 %**) pour un
vrai extrait de transcrit humain, en n'utilisant que des variables, des
types de données et des méthodes de chaînes.
